### 1.:

We consider a single test point $x_0$, held fixed, with

\begin{align*}
y_0 = f(x_0) + \varepsilon_0, \qquad \varepsilon_0 \sim \mathcal{N}(0,\sigma^2)
\end{align*}

The prediction $\tilde{y}_0 = \tilde{y}_0(\mathcal{L})$ is produced by a model fitted on a
training set $\mathcal{L}$, and $x_0$ is not part of $\mathcal{L}$. There are therefore two
independent sources of randomness: the draw of the training set $\mathcal{L}$, and the
noise $\varepsilon_0$ on the test point. All expectation values below are taken
over both. The point $x_0$ itself is held fixed, so $f_0 \equiv f(x_0)$ is a
constant. [Machine Learning and the Physical Sciences, Morten Hjorth-Jensen (Springer, 2026), Sec.~2.10, Eqs.~(2.49)--(2.52)]

We define $\mu_0 = \mathbb{E}[\tilde{y}_0]$, the expectation of the prediction over training sets.

We start with the term $(y_0 - \tilde{y}_0)$, looking at a single component:

\begin{align*}
y_0 - \tilde{y}_0
  &= \underbrace{(f_0 + \varepsilon_0)}_{=\,y_0} - \mu_0 + \mu_0 - \tilde{y}_0 \\[3mm]
  &= \underbrace{(f_0 - \mu_0)}_{=\,A}
   + \underbrace{(\mu_0 - \tilde{y}_0)}_{=\,B}
   + \underbrace{\varepsilon_0}_{=\,C}
\end{align*}

where $A$ is constant, $B$ is random through the choice of training set,
and $C$ is random through the noise.

Squaring and taking the expectation value:

\begin{align*}
\mathbb{E}\left[(y_0 - \tilde{y}_0)^2\right]
= \mathbb{E}\left[A^2 + B^2 + C^2 + 2AB + 2AC + 2BC\right]
\end{align*}


Since the expectation value is linear, we can treat each term separately:

\begin{align*}
\mathbb{E}\left[(y_0-\tilde{y}_0)^2\right]
= \mathbb{E}[A^2] + \mathbb{E}[B^2] + \mathbb{E}[C^2]
+ 2\mathbb{E}[AB] + 2\mathbb{E}[AC] + 2\mathbb{E}[BC]
\end{align*}

Now if we look at each expected value:

\begin{align*}
\mathbb{E}[A^2]
  &= A^2 = (f_0-\mu_0)^2
  && \text{since } A \text{ is constant} \\[2mm]
\mathbb{E}[B^2]
  &= \mathbb{E}\left[(\mu_0-\tilde{y}_0)^2\right]
   = \mathbb{E}\left[(\tilde{y}_0-\mu_0)^2\right]
  && \text{squaring removes the sign, and for later, } \mu_0 = \mathbb{E}[\tilde{y}_0] \\[2mm]
\mathbb{E}[C^2]
  &= \mathbb{E}[\varepsilon_0^2] = \sigma^2
  && \text{since } \mathbb{E}[\varepsilon_0]=0
   \Rightarrow \mathbb{E}[\varepsilon_0^2]=\mathrm{var}[\varepsilon_0]=\sigma^2
\end{align*}

and the three cross terms all vanish:

\begin{align*}
2\mathbb{E}[AB]
  &= 2A\,\mathbb{E}[B] = 2A\left(\mu_0-\mathbb{E}[\tilde{y}_0]\right) = 0
  && A \text{ constant, and } \mathbb{E}[\tilde{y}_0]=\mu_0 \text{ by definition} \\[2mm]
2\mathbb{E}[AC]
  &= 2A\,\mathbb{E}[\varepsilon_0] = 0
  && A \text{ constant, and the noise has zero mean} \\[2mm]
2\mathbb{E}[BC]
  &= 2\,\mathbb{E}[B]\,\mathbb{E}[\varepsilon_0] = 0
  && \text{the test noise is independent of the training set}
\end{align*}

Putting it all together:

\begin{align*}
\mathbb{E}\left[(y_0-\tilde{y}_0)^2\right]
= \underbrace{(f_0-\mu_0)^2}_{\mathrm{Bias}^2[\tilde{y}_0]}
+ \underbrace{\mathbb{E}[(\tilde{y}_0-\mu_0)^2]}_{\mathrm{Var}[\tilde{y}_0]}
+ \underbrace{\sigma^2}_{\text{noise}}
+ 0 + 0 + 0
\end{align*}

Since we have only looked at a single point, we can now average over the test
points $i = 0, 1, \dots, n-1$:

\begin{align*}
\frac{1}{n}\sum_{i=0}^{n-1}\mathbb{E}\left[(y_i-\tilde{y}_i)^2\right]
  &= \frac{1}{n}\sum_{i=0}^{n-1}\left(f_i-\mu_i\right)^2
   + \frac{1}{n}\sum_{i=0}^{n-1}\mathbb{E}\left[(\tilde{y}_i-\mu_i)^2\right]
   + \sigma^2
\end{align*}

The cost function of the exercise, $C=\frac1n\sum_{i=0}^{n-1}(y_i-\tilde y_i)^2$, is a random variable: it depends on the training set $\mathcal L$ and on the noise $\boldsymbol\varepsilon$. The bias-variance decomposition is a statement about its expectation. By linearity of the expectation,
$$
\mathbb E_{\mathcal L,\varepsilon}[C] = \frac1n\sum_{i=0}^{n-1}\mathbb E_{\mathcal L,\varepsilon}\big[(y_i-\tilde y_i)^2\big],
$$
which is the left-hand side above. This is what is written $\mathbb E[(\boldsymbol y-\tilde{\boldsymbol y})^2]$ in the exercise, where $\mathbb E$ is understood as the expectation over $\mathcal L$ and $\boldsymbol\varepsilon$ (in practice estimated by sample averages, e.g. over bootstrap resamples), not as the average over the $n$ points alone.
Likewise $\mathbb E[\tilde{\boldsymbol y}]$ is the vector with components $\mu_i=\mathbb E_{\mathcal L}[\tilde y_i]$, one mean per test point. Inserting the single-point result for each $i$ gives

\begin{align*}
\mathbb E\big[(\boldsymbol y-\tilde{\boldsymbol y})^2\big]
&= \frac1n\sum_i (f_i-\mu_i)^2 + \frac1n\sum_i \mathbb E_{\mathcal L}\big[(\tilde y_i-\mu_i)^2\big] + \frac1n\sum_i\sigma^2 \\
&= \underbrace{\frac1n\sum_i \big(f_i-\mathbb E[\tilde y_i]\big)^2}_{\mathrm{Bias}^2[\tilde{\boldsymbol y}]}
 + \underbrace{\frac1n\sum_i \mathbb E_{\mathcal L}\big[(\tilde y_i-\mathbb E[\tilde y_i])^2\big]}_{\mathrm{var}[\tilde{\boldsymbol y}]}
 + \sigma^2 .
\end{align*}

The bias term does not depend on $\mathcal L$ or $\varepsilon$ ($\boldsymbol f$ is not stochastic), so there the expectation only acts as the average over $i$. In the variance term, the expectation over $\mathcal L$ is understood in the exercise's expression $\frac1n\sum_i(\tilde y_i-\mathbb E[\tilde{\boldsymbol y}])^2$, and in practice it is estimated by a sample average (the bootstrap).

### 2.:


For me, the easiest way to visualize the absorption of $\sigma^2$ into the bias
term is by starting at our first equation above, where we had

\begin{align*}
y_0 - \tilde{y}_0
= \underbrace{(f_0 - \mu_0)}_{=\,A}
+ \underbrace{(\mu_0 - \tilde{y}_0)}_{=\,B}
+ \underbrace{\varepsilon_0}_{=\,C}
\end{align*}

If we now move $\varepsilon_0$ behind $f_0$, we get, by definition of $y_0$,

\begin{align*}
y_0 - \tilde{y}_0
  &= \big(\underbrace{f_0 + \varepsilon_0}_{=\,y_0} - \mu_0\big)
   + (\mu_0 - \tilde{y}_0) \\[3mm]
  &= \underbrace{(y_0 - \mu_0)}_{=\,a}
   + \underbrace{(\mu_0 - \tilde{y}_0)}_{=\,b}
\end{align*}

We can now resume the derivation the same way as we did previously:

\begin{align*}
\mathbb{E}\left[(a+b)^2\right]
= \mathbb{E}[a^2] + \mathbb{E}[b^2] + 2\mathbb{E}[ab]
\end{align*}

\begin{align*}
\mathbb{E}[a^2]
  &= \mathbb{E}\left[(y_0-\mu_0)^2\right]
   = \mathbb{E}\left[\big((f_0+\varepsilon_0)-\mu_0\big)^2\right] \\[2mm]
  &= (f_0-\mu_0)^2 + 2(f_0-\mu_0)\,\mathbb{E}[\varepsilon_0] + \mathbb{E}[\varepsilon_0^2] \\[2mm]
  &= (f_0-\mu_0)^2 + 0 + \sigma^2 \\[2mm]
  &= (f_0-\mu_0)^2 + \sigma^2
\end{align*}

\begin{align*}
\mathbb{E}[b^2]
  &= \mathbb{E}\left[(\mu_0-\tilde{y}_0)^2\right]
   = \mathbb{E}\left[(\tilde{y}_0-\mu_0)^2\right]
  && \text{same as the previous } B \\[2mm]
\mathbb{E}[2ab]
  &= 2\,\mathbb{E}[a]\,\mathbb{E}[b] = 0
  && \text{by independence, and } \mathbb{E}[b]=0
\end{align*}

So we get

\begin{align*}
\mathbb{E}\left[(y_0-\tilde{y}_0)^2\right]
= \underbrace{(f_0-\mu_0)^2 + \sigma^2}_{\text{measured bias}^2}
+ \underbrace{\mathbb{E}\left[(\tilde{y}_0-\mu_0)^2\right]}_{\mathrm{var}[\tilde{y}_0]}
\end{align*}

So we see that the bias computed in the code is not the true squared
bias, but the true squared bias plus the noise variance,
$(f_0-\mu_0)^2 + \sigma^2$. The consequence is that the measured bias has a florr at $\sigma^2$ in expectation, however well the model fits.

### 3.:

### Interpretation of the three terms

The **squared bias** measures how far the average prediction lies from the true
function. It reflects the model family we chose: a low-degree polynomial cannot
reproduce the shape of the Runge function, and collecting more data will not
repair that. It therefore falls as the degree increases, and depends only
weakly on the number of data points.

The **variance** measures how much the prediction at a fixed point moves when
the training set is redrawn. A flexible model adapts to the particular noise
realisation it is handed, so its predictions differ from one training set to
the next. It grows with the degree, and shrinks as more data stabilise each fit.

The **irreducible error** $\sigma^2$ is the variance of the noise itself.
Nothing in the data allows a model to predict $\varepsilon_0$, so $\sigma^2$ is
a floor beneath the test error for any model whatsoever.

Since the bias falls and the variance grows with complexity, their sum has a
minimum at some intermediate degree. Locating that minimum is the purpose of
the analysis that follows.

### Over which randomness the expectation values are taken

Throughout the derivation, $\mathbb{E}[\cdot]$ runs over the two independent
sources of randomness introduced at the start: the draw of the training set
$\mathcal{L}$, and the noise $\varepsilon_0$ on the test point.

The bootstrap realises only the first of these directly. The split into
training and test data is made once, $y_{\text{test}}$ is drawn once and held
fixed, and only the training set is varied, by resampling it. The average over
$\varepsilon_0$ is replaced by the average over the $n$ test points. This is
why the noise at a given point is never averaged away, and why it remains
inside the measured bias term derived above.
